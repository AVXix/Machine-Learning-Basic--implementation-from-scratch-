Use NumPy library to complete the following tasks:

```python
import numpy as np
import pandas as pd
```

```python
# 1. Create two linear arrays with 20 random elements each ranging from 1 to 100
a = np.random.randint(1, 101, 20)
b = np.random.randint(1, 101, 20)

print("1. Linear arrays:")
print("Array a:", a)
print("Array b:", b)
print()
```

```python
# 2. Change linear arrays to two 5x4 matrices
matrix_a = a.reshape(5, 4)
matrix_b = b.reshape(5, 4)


print("5x4 Matrix A:")
print(matrix_a)
print("5x4 Matrix B:")
print(matrix_b)
print()
```

```python
# 3. Perform scalar multiplication with matrix a and b
scalar = 2
scalar_mult_a = scalar * matrix_a
scalar_mult_b = scalar * matrix_b

print("Scalar multiplication of Matrix A:")
print(scalar_mult_a)
print("Scalar multiplication of Matrix B:")
print(scalar_mult_b)
print()
```

```python
# 4. Modify matrix b and perform matrix multiplication
matrix_b_modified = matrix_b.T  # Transpose to make it 4x5

# Matrix multiplication
matrix_c = np.matmul(matrix_a, matrix_b_modified)

print("4. Matrix multiplication:")
print("Modified Matrix B (4x5):")
print(matrix_b_modified)
print("Matrix C (result of A × B_modified):")
print(matrix_c)
print()
```

```python
# 5. Print only the column 2, 3 of matrix c. Print the whole matrix using slicing technique.
print("Columns 2 and 3 of Matrix C:")
print(matrix_c[:, 1:3]) # Python uses 0-based indexing, so columns 2 and 3 are at index 1 and 2

print("\nWhole Matrix C using slicing:")
print(matrix_c[:, :]) # Using slicing to print the whole matrix
```

```python
# 6. Find the largest element and its index
max_value = np.max(matrix_c)
max_index = np.unravel_index(np.argmax(matrix_c), matrix_c.shape)

print(f"Largest element: {max_value}")
print(f"Index of largest element: {max_index}")
print()
```

```python
# 7. Convert matrix c to one-dimensional array keeping original unchanged
matrix_c_flat = matrix_c.flatten()

print("Flattened 1D array:")
print(matrix_c_flat)
print()
```

Use Matplotlib library to complete the

```python
import matplotlib.pyplot as plt
```

```python
temperature_dhaka = [25,34,21,45,28,6,43,18,7,2]
humidity_dhaka = [28, 25,29,20, 26, 50, 19, 29, 52, 55]
```

Generate a scatter plot graph for temperature_dhaka vs humidity_dhaka using (*) as the
marker in color red.

```python
plt.scatter(temperature_dhaka, humidity_dhaka, marker='*', color='red')
plt.xlabel("Temperature (°C)")
plt.ylabel("Humidity (%)")
plt.title("Temperature vs Humidity in Dhaka")
plt.grid(True)
plt.show()
```

Now do a line plot for study_hours vs marks with figure size (12,8)

```python
# Sample data for study hours and marks
study_hours = np.array([2, 3, 5, 6, 7, 8, 10, 11, 12, 14])
marks = np.array([50, 60, 70, 75, 80, 85, 90, 92, 95, 98])

# Set the figure size
plt.figure(figsize=(12, 8))

# Create the line plot
plt.plot(study_hours, marks, marker='o', linestyle='-')

# Add labels and title
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks")
plt.grid(True)

# Show the plot
plt.show()
```

```python
# Calculate the average score for each subject
average_scores = DataFrames[['Physics', 'Chemistry', 'Math', 'Bilogy']].mean()

# Create a horizontal bar graph
plt.figure(figsize=(10, 6))
average_scores.plot(kind='barh', color=['skyblue', 'lightgreen', 'salmon', 'gold'])

# Add labels and title
plt.xlabel("Average Score")
plt.ylabel("Subject")
plt.title("Average Subject Scores")
plt.grid(axis='x')

# Show the plot
plt.show()
```

```python
# Calculate the average score for each subject
average_scores = DataFrames[['Physics', 'Chemistry', 'Math', 'Bilogy']].mean()

# Create a vertical bar graph
plt.figure(figsize=(10, 6))
average_scores.plot(kind='bar', color=['skyblue', 'lightgreen', 'salmon', 'gold'])

# Add labels and title
plt.xlabel("Subject")
plt.ylabel("Average Score")
plt.title("Average Subject Scores")
plt.grid(axis='y')

# Show the plot
plt.show()
```

```python
# Calculate the average score for each subject
average_scores = DataFrames[['Physics', 'Chemistry', 'Math', 'Bilogy']].mean()

# Create a figure and subplots
fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(16, 12))

# Scatter plot
axes[0, 0].scatter(temperature_dhaka, humidity_dhaka, marker='*', color='red')
axes[0, 0].set_xlabel("Temperature (°C)")
axes[0, 0].set_ylabel("Humidity (%)")
axes[0, 0].set_title("Temperature vs Humidity in Dhaka")
axes[0, 0].grid(True)

# Line plot
axes[0, 1].plot(study_hours, marks, marker='o', linestyle='-')
axes[0, 1].set_xlabel("Study Hours")
axes[0, 1].set_ylabel("Marks")
axes[0, 1].set_title("Study Hours vs Marks")
axes[0, 1].grid(True)

# Horizontal bar graph
average_scores.plot(kind='barh', color=['skyblue', 'lightgreen', 'salmon', 'gold'], ax=axes[1, 0])
axes[1, 0].set_xlabel("Average Score")
axes[1, 0].set_ylabel("Subject")
axes[1, 0].set_title("Average Subject Scores (Horizontal)")
axes[1, 0].grid(axis='x')

# Vertical bar graph
average_scores.plot(kind='bar', color=['skyblue', 'lightgreen', 'salmon', 'gold'], ax=axes[1, 1])
axes[1, 1].set_xlabel("Subject")
axes[1, 1].set_ylabel("Average Score")
axes[1, 1].set_title("Average Subject Scores (Vertical)")
axes[1, 1].grid(axis='y')

# Adjust layout to prevent overlapping titles
plt.tight_layout()

# Show the plot
plt.show()
```
